- Ich **bereue** dass ich nicht früher angefangen hatte, Deutsch zu lernen. In Nachhinhein, würde ich es heute anders machen und würde mich schon früher mit Deutsch beschäftigen. Ich kann mich sehr gut und spontan auf Englisch **ausdrücken**, weil ich seit meiner Schulzeit Englisch gelernt und viel geübt habe. Meine Vorlesungen in der Universität z.B. wurden alle auf Englisch gehalten.
- Wenn ich versuche Deutsch zu sprechen, muss ich immer noch alles in meinem Kopf zuerst **übersetzen**. Das führt zu Schwerigtkeiten beim Gespräche und ich muss nach ein paar Minute abbrechen weil das Gespräch unangenehm fühlt. Es ist manchmal schwer deinen eigenen Fortschritt zu erkennen. Wie man sagt: Rom wurde auch nicht an einem Tag erbaut. 
- Ich habe zwischendurch Persich gelernt. Die Persiche Sprache finde ich sehr toll und sie klingt für mich wunderschön. Die Wortschatz **kommit** mir als ein Mutterspracher von Urdu sehr **bekannt vor**. 
- Ich habe die Erfahrung gemacht dass viele Leute **geben auf** bevor sie das Niveau B1 erreicht haben. Doch B1 ist das Niveau wo es meiner Meinung nach Spaß macht. Man lernt die Sprache, die man sofort im Alltag benutzen kann und genau das finde ich **aufregend** und besonders beim Arbeitsleben unerlässlich.
- Die Ausländer müssen in Deutschland eine **Arbeitserlaubnis** beantragen um arbeiten zu dürfen. Ohne diese Erlaubnis, die von der Bundesagentur für Arbeit erteilt wird, dürfen sie keine Arbeit oder Erwerbstätigkeit aufnehmen.
- Ich bin **mitterweile** nach Berlin gezogen. Ich wird noch eine Weile dauern bis ich mich hier in Berlin einleben können.
- Ein Freund von mir hat während der Probezeit ein Monate lang Urlaub. Ich würde persönlich sowas nie riskieren. So könnte man schell **gekündigt** werden.

## Use of 'auf' und 'um' when it comes to numbers and percentages

The basic logic is:
- um = the amount of change → by => Think of um as answering: Um wie viel? (By how much?)

- auf = the final value/result → to => Think of auf as answering: Auf welchen Wert? (To what final value?)

1. Um den Einzelhandel anzukurbeln hat die Regierung die Mehrwertsteur auf 15% gesenkt. Der neue Steuersatz wird 

To stimulate the retail industry the government has reduced the VAT to 15% (from a previously higher amount e.g. 19%)

2. Ab dem nächten Monat werde die Miete werde um 100 Euro erhöht.

3. Wegen des herrschenden Krieges in der Ukraine und die damit gebundene Gaslieferungssperre von Russland, sind die Heizungskosten überall stark gestiegen.
